```python
import pandas as pd

fichier = "reponses.xlsx"

df = pd.read_excel(fichier)

df = df.dropna(how="all")

print("Nombre de participants :", len(df))
print("Nombre de colonnes :", len(df.columns))

print("\nAperçu des réponses :")
display(df.head())

print("\nQuestions disponibles :")
for i, question in enumerate(df.columns, start=1):
    print(i, question)
print("Nombre de participants :", len(df))
print("Nombre de colonnes :", len(df.columns))
print("Valeurs manquantes :", df.isnull().sum().sum())
```

```python
questions = df.iloc[:, 1:23].copy()

print("Nombre de questions :", questions.shape[1])
print("\nTypes des colonnes :")
print(questions.dtypes)

display(questions.head())
```

```python
codes = {
    "Jamais": 1,
    "Rarement": 2,
    "Parfois": 3,
    "Souvent": 4,
    "Systématiquement": 5
}

for numero in [2, 4, 9, 19]:
    colonne = questions.columns[numero - 1]
    questions[colonne] = questions[colonne].map(codes)

print("Nombre de questions :", questions.shape[1])
print("Valeurs manquantes après conversion :",
      questions.isnull().sum().sum())

display(questions.head())
```

```python
q1 = questions.iloc[:, 0]

moyenne = q1.mean()
ecart_type = q1.std(ddof=1)
effectif = q1.count()

print("Moyenne Q1 :", round(moyenne, 4))
print("Écart-type Q1 :", round(ecart_type, 4))
print("Effectif Q1 :", effectif)
```

```python
from scipy import stats
import math

t = stats.t.ppf(0.975, effectif - 1)

marge = t * ecart_type / math.sqrt(effectif)

ic_min = moyenne - marge
ic_max = moyenne + marge

print("IC95 inférieur :", round(ic_min, 4))
print("IC95 supérieur :", round(ic_max, 4))
```

```python
from scipy import stats
import math

resultats = []

for i in range(22):
    valeurs = questions.iloc[:, i].dropna()

    n = len(valeurs)
    moyenne = valeurs.mean()
    ecart_type = valeurs.std(ddof=1)

    t = stats.t.ppf(0.975, n - 1)
    marge = t * ecart_type / math.sqrt(n)

    resultats.append({
        "Question": "Q" + str(i + 1),
        "Moyenne": moyenne,
        "Ecart_type": ecart_type,
        "Effectif": n,
        "IC95_Min": moyenne - marge,
        "IC95_Max": moyenne + marge,
        "Score_%": moyenne / 5 * 100
    })

tableau_stats = pd.DataFrame(resultats)

display(tableau_stats.round(4))
```

```python
questions_risques = [11, 12, 13, 14, 1, 2, 3]

for numero in questions_risques:
    ligne = tableau_stats.loc[
        tableau_stats["Question"] == "Q" + str(numero)
    ]
    print("Q" + str(numero),
          ":", round(ligne["Score_%"].iloc[0], 2), "%")
```

```python
def score(numero):
    ligne = tableau_stats[
        tableau_stats["Question"] == "Q" + str(numero)
    ]
    return ligne["Score_%"].iloc[0] / 100

x1 = score(11)
y1 = score(12)

x2 = score(13)
y2 = score(14)

x3 = (score(1) + score(2)) / 2
y3 = score(3)

print("R1 :", round(x1 * 100, 2), "% |",
      round(y1 * 100, 2), "%")
print("R2 :", round(x2 * 100, 2), "% |",
      round(y2 * 100, 2), "%")
print("R3 :", round(x3 * 100, 2), "% |",
      round(y3 * 100, 2), "%")
```

```python
risques = [
    ("R1", x1, y1),
    ("R2", x2, y2),
    ("R3", x3, y3),
    ("R5", x5, y5),
    ("R6", x6, y6),
    ("R7", x7, y7),
    ("R8", x8, y8),
    ("R9", x9, y9),
    ("R10", x10, y10),
    ("R11", x11, y11),
    ("R12", x12, y12),
]

for nom, xi, yi in risques:
    ci = xi * yi

    if xi >= 0.5 and yi >= 0.5:
        zone = "IV"
    elif xi < 0.5 and yi >= 0.5:
        zone = "II"
    elif xi >= 0.5 and yi < 0.5:
        zone = "III"
    else:
        zone = "I"

    print(nom, "| Criticité :", round(ci * 100, 2),
          "% | Zone :", zone)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

for nom, xi, yi in risques:
    plt.scatter(xi * 100, yi * 100)
    plt.annotate(nom, (xi * 100, yi * 100),
                 xytext=(5, 5),
                 textcoords="offset points")

plt.axvline(50, color="gray", linestyle="--")
plt.axhline(50, color="gray", linestyle="--")

plt.xlim(0, 100)
plt.ylim(0, 100)

plt.xlabel("Occurrence Xi (%)")
plt.ylabel("Impact Yi (%)")
plt.title("Cartographie des risques de fraude aux examens")
plt.grid(alpha=0.3)

plt.show()
```

```python
import pandas as pd

classement = pd.DataFrame(
    risques,
    columns=["Risque", "Xi", "Yi"]
)

classement["Ci"] = classement["Xi"] * classement["Yi"]

classement = classement.sort_values(
    by="Ci",
    ascending=False
)

display(classement.round(4))
```

```python
def indice_x(x):
    if x < 0.25:
        return 1
    elif x < 0.50:
        return 2
    elif x < 0.75:
        return 3
    return 4

def indice_y(y):
    if y < 0.20:
        return 1
    elif y < 0.40:
        return 2
    elif y < 0.60:
        return 3
    elif y < 0.80:
        return 4
    return 5

print("R1 :", indice_x(x1), indice_y(y1))
print("R7 :", indice_x(x7), indice_y(y7))
print("R9 :", indice_x(x9), indice_y(y9))
```

```python
classement["Indice_Xi"] = classement["Xi"].apply(indice_x)
classement["Indice_Yi"] = classement["Yi"].apply(indice_y)

def determiner_zone(x, y):
    if x >= 0.5 and y >= 0.5:
        return "IV"
    elif x < 0.5 and y >= 0.5:
        return "II"
    elif x >= 0.5 and y < 0.5:
        return "III"
    else:
        return "I"

classement["Zone"] = classement.apply(
    lambda ligne: determiner_zone(ligne["Xi"], ligne["Yi"]),
    axis=1
)

display(classement.round(4))
```
